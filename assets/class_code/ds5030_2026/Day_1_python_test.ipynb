## 2. Can you read a data file and draw a picture?

`cville_cars.csv` is 92 used cars scraped from Charlottesville Craigslist. It sits next to this notebook
in the repository.

```python
import pandas as pd
import matplotlib.pyplot as plt

cars = pd.read_csv("cville_cars.csv")
print("rows, columns:", cars.shape)

cars[["title", "price", "year", "miles"]].head()
```

```python
cars["price"].plot(kind="hist", bins=20, title="Charlottesville used-car prices")
plt.xlabel("price ($)")
plt.show()
```

Most of the cars are cheap, with a long tail of expensive ones. That lopsided shape is a large part of
why this course spends September on distributions and densities rather than jumping to averages.
