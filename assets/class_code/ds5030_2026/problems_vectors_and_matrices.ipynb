**Problem 1.** Consider the vectors
$$
x = \left[ \begin{array}{cc} 2 & -1 \end{array} \right]
\quad \text{and} \quad
y = \left[ \begin{array}{c} 3 \\ 6 \end{array} \right].
$$

Compute the following, using broadcasting where necessary:

- a. $x - y^{\top}$
- b. $2x^{\top} - \frac{1}{3}y$
- c. $x - y$
- d. $x^{\top} \odot y$
- e. $x \odot y$

**Problem 2.** Consider the matrices
$$
A = \left[ \begin{array}{cc} 2 & 1 \\ 0 & 3 \end{array} \right],
\quad
B = \left[ \begin{array}{cc} 1 & 4 \\ -2 & 5 \end{array} \right],
\quad \text{and} \quad
C = \left[ \begin{array}{cc} 2 & -1 \\ 0 & 4 \end{array} \right].
$$
The **transpose** of a matrix $D$, $D^{\top}$, is the matrix with $D_{ij}$ swapped with $D_{ji}$, swapping rows with columns.

- a. Is $A-B = -B+A$? Compute both sides.
- b. Is $A \cdot B = B \cdot A$? Compute both products.
- c. Is $(A \cdot B)^{\top} = A^{\top} \cdot B^{\top}$? Compute both sides. What about $B^{\top} \cdot A^{\top} $?
- d. Is $(A \odot B) \odot C = A \odot (B \odot C)$? Compute both sides.

**Problem 6.** Consider the vectors
$$
x = \left[ \begin{array}{c} 1 \\ 1 \\ 0 \end{array} \right],
\quad
y = \left[ \begin{array}{c} 1 \\ -1 \\ 0 \end{array} \right],
\quad \text{and} \quad
z = \left[ \begin{array}{c} 1 \\ 0 \\ 1 \end{array} \right].
$$

- a. Are any of these vectors orthogonal to one another?
- b. The **Gram-Schmidt process** takes a set of vectors and produces an orthogonal set. It works like this:
  1. Let $u_1 = x$.
  2. Set $u_2 = y - \frac{y \cdot x}{x \cdot x}x$.
  3. Set $u_3 = z - \frac{z \cdot u_1}{u_1 \cdot u_1}u_1 - \frac{z \cdot u_2}{u_2 \cdot u_2}u_2$.
  Apply this process to $x$, $y$, and $z$.
- c. Show that $u_1$, $u_2$, and $u_3$ are orthogonal.

**Problem 7.** Imagine a dataset of pairs $(y_i,x_i)$ with
$$
y = \left[ \begin{array}{c} 2 \\ 3 \\ 2 \end{array} \right]
\quad \text{and} \quad
x = \left[ \begin{array}{c} 1 \\ 2 \\ 3 \end{array} \right].
$$
Consider the predictive model
$$
\hat{y}_i = bx_i,
$$
so the error is
$$
e_i = y_i - \hat{y}_i = y_i - bx_i.
$$

- a. Show that the sum of squared error can be written as
$$
\mathrm{SSE}(b) = (y-\hat y) \cdot (y-\hat y).
$$
- b. Compute the optimal slope
$$
b^* = \frac{x \cdot y}{x \cdot x}.
$$
- c. Compute the error vector at $b^*$ and show that $e \cdot x = 0$.
- d. Is the average error equal to $0$ in this example? What modification of the model would force the average error to be $0$?

**Problem 9.** The sample covariance of two variables $X_j = [x_{1j}, x_{2j}, ..., x_{nj}]^{\top}$ and $X_k =  [x_{1k}, x_{2k}, ..., x_{nk}]^{\top}$ can be written as
$$
\hat{c}_{jk} = \frac{\sum_{i=1}^n (x_{ij}-m(X_j)) (x_{ik} - m(X_k))}{n}
$$
and if $j = k$, then we get the sample variance,
$$
\hat{c}_{jj} = \frac{ \sum_{i=1}^n (x_{ij}-m(X_j))^2}{n}.
$$

Let
$$
X_1 = \left[ \begin{array}{c} 1 \\ 2 \\ 3 \end{array} \right]
\quad \text{and} \quad
X_2 = \left[ \begin{array}{c} 2 \\ 0 \\ 4 \end{array} \right] \quad \text{and} \quad X_3 = \left[ \begin{array}{c} 3 \\ 1 \\ 2 \end{array} \right]
$$

- a. Mean-center the variables and how how the sample covariances $\hat c_{12}$, $\hat c_{23}$, and $\hat c_{13}$ and sample variances $\hat{c}_{11}$, $\hat{c}_{22}$, and $\hat{c}_{33}$ can be written as an inner product calculations. Compute these values.
- b. The **variance co-variance matrix** is
$$
\hat{C} = \left[ \begin{array}{cccc} 
\hat{c}_{11} & \hat{c}_{12} & \dots & \hat{c}_{1L} \\
\hat{c}_{21} & \hat{c}_{22} & \dots & \hat{c}_{2L} \\
\vdots & \vdots & \ddots & \vdots \\
\hat{c}_{L1} & \hat{c}_{L2} & \dots & \hat{c}_{LL} \\
\end{array} \right]
$$
Let $Z$ be the matrix whose rows are the mean-centered observation vectors. Compute $Z$ and then compute
$$
\hat C = \frac{Z^{\top}Z}{3}.
$$
Show that the resulting matrix calculation matches your hand-by-hand calculations from part a.
