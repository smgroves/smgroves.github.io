## Transcendentals and Gaussian Function

## Building the Exponential Function
- Imagine we start with the constant function $e_0(x)=1$, and integrate it, to get
$$
e_1(x) = \int e_0(x)dx = \int 1 dx = 1 + x
$$
- Let's integrate again:
$$
e_2(x) = \int e_1(x) dx = \int 1 + x dx = 1 + x + \frac{x^2}{2}
$$
- Keep going:
$$
e_k(x) = \int e_{k-1}(x) dx = \int 1 + x + ... + \frac{x^{k-1}}{(k-1)!} dx = 1 + x + \frac{x^2}{2} + ... + \frac{x^k}{k!}
$$
- If we do this forever, what do we get?
$$
e(x) = 1 + x + \frac{x^2}{2} + ... + \frac{x^k}{k!} + ...
$$
- This is called **the exponential function**

## $e^x$ is non-negative
- Notice that $e^0 = 1 + \frac{1}{2} + \frac{1}{6} + ... \frac{1}{k!} + ... = e >0$
- If $x>0$, then $e^x > 0 $, since we have a positive base to a positive exponent, so it must be non-negative
- If $x<0$, then $e^x = 1/e^{-x} >0$, and we 1 divided by a positive base to a positive exponent, so it must be non-negative

## $D_x e^x = e^x$ and $D_x e^x >0$
- Notice:
$$
\begin{alignat*}{2}
e^x &=& 1 + x + \frac{x^2}{2} + ... + \frac{x^k}{k!} + ... \\
D_x e^x &=& 0 + 1 + x + \frac{x^2}{2} + ... + \frac{x^{k-1}k}{k!} +\frac{x^{k}(k+1)}{(k+1)!} + ... \\
&=& e^x
\end{alignat*}
$$
- Since $e^x >0$ and $ D_x e^x = e^x$, this implies $D_x e^x >0$
- So the exponential function is a **fixed-point** of differentiation and integration

## $e^{x+y} = e^x e^y$
- This is a consequence of the binomial theorem,
$$
(a+b)^n = \sum_{k=0}^n \frac{k!}{n!(n-k)!} a^k b^{n-k}
$$
- Let's write it out:
$$
\begin{alignat*}{2} e^{x+y} &= \sum_{n=0}^{\infty} \dfrac{(x+y)^n}{n!} && \text{(Definition)} \\ 
&= \sum_{n=0}^{\infty} \sum_{k=0}^{n} \frac{1}{n!} \frac{n!}{k!(n-k)!} x^k y^{n-k} \quad && \text{(Binomial expansion; notice } k \text{ stops at } n) \\ 
&= \sum_{n=0}^{\infty} \sum_{k=0}^{n} \frac{x^k}{k!} \frac{y^{n-k}}{(n-k)!} && \text{(The } n! \text{ terms cancel)} \\ 
&= \sum_{k=0}^{\infty} \frac{x^k}{k!} \sum_{n=k}^{\infty} \frac{y^{n-k}}{(n-k)!} && \text{(Swap the order of summation region: } 0 \le k \le n < \infty) \\ 
&= \sum_{k=0}^{\infty} \frac{x^k}{k!} \sum_{m=0}^{\infty} \frac{y^m}{m!} && \text{(Substituting } m = n-k\text{, shifting the boundary to } 0) \\ 
&= e^x e^y && \text{(Definition)} 
\end{alignat*}
$$

## Example: Softmax
- We often have a finite number of labels, $\ell_1, \ell_2, ..., \ell_k$ and intensities $v_1 \ge v_2 \ge ... \ge v_K$
- We want to randomly select a label, in such a way that labels with larger intensities are more likely
- The most popular way to do this is called **softmax**:
$$
p[\ell_j | v ] = \dfrac{e^{v_j/s}}{\sum_{k = 1}^K e^{v_k/s}}
$$
- Notice, these are really probabilites because you can sum over all the labels, and the numerator and denominator cancel out
- It is called softmax because as $s \rightarrow 0$, you pick the label with the highest intensity with probability 1:
$$
p[\ell_1 | v ] = \dfrac{1}{1 + \sum_{k = 2}^K e^{(v_k-v_1)/s}}, \quad p[\ell_j | v ] = \dfrac{e^{(v_j-v_1)/s}}{1 + \sum_{k = 2}^K e^{(v_k-v_1)/s}}
$$
and 
$$
\lim_{s \rightarrow 0} \dfrac{1}{1+\sum_{k = 2}^K e^{(v_k-v_1)/s}} = \frac{1}{1}, \quad \lim_{s \rightarrow 0} \dfrac{e^{(v_j-v_1)/s}}{1 + \sum_{k = 2}^K e^{(v_k-v_1)/s}} = \frac{0}{1}
$$
and if $s \rightarrow \infty$, $e^{y/s} \rightarrow e^0 = 1$, and
$$
\lim_{s \rightarrow \infty} \dfrac{1}{1+\sum_{2 = 1}^K e^{(v_k-v_1)/s}} = \frac{1}{K}, \quad \lim_{s \rightarrow \infty} \dfrac{e^{(v_j-v_1)/s}}{1 + \sum_{k = 2}^K e^{(v_k-v_1)/s}} = \frac{1}{K}
$$
which is the uniform mass function.

## Example: The Gaussian Kernel
- The exponential ($1-e^{\lambda t}$), logistic ($1/(1+e^{-x})$) and standard normal ($e^{-x^2/2}/\sqrt{2 \pi}$) distributions all use the exponential function as a fundamental ingredient
- In particular, the **Gaussian kernel**, the **standard normal density** is important:
$$
\phi(z) = \frac{1}{\sqrt{2\pi}} \exp \left( \frac{-z^2}{2} \right)
$$
- This is the equation for "the bell curve"

## Properties of the Gaussian Kernel

- It is a density: $\int_{-\infty}^\infty e(-z^2/2)dz = \sqrt{2\pi}$, so $\int_{-\infty}^\infty \phi(z) dz = 1$
- Since $-z^2$ is negative and $e^x$ is increasing, $\phi(z)$ decreases as you move away from 0: It is maximized at 0 with a value of $1/\sqrt{2\pi}$ and trends quickly to zero as $z^2$ becomes large
- It must be non-negative, because $\exp(w)\ge 0$
- The log equals $\log(\phi(z)) = \frac{-z^2}{2} - \log(\sqrt{2\pi})$


## The Natural Logarithm
- The inverse of the exponential is defined as the (natural) logarithm: $\log(e^x) = x$
- We want to understand, for other values $\log(y)$, how does it behave?
- First, since $e^x \ge 0$, $\log()$ is only defined over positive real numbers, not negative real numbers
- As $y \rightarrow 0$, $y = e^{x}$ must satisfy $x \rightarrow -\infty$, so that $\lim_{y \rightarrow 0} \log(y) = -\infty$

## $\log'(y) = 1/y$
Notice that:
$$
\begin{alignat*}{2}
\log(e^x) &=& x \quad (\text{Definition} )\\
\log'(e^x)e^x &=& 1 \quad (\text{Chain rule} )\\
\log'(e^x) &=& \frac{1}{e^x} \quad (\text{Rearrange} )\\
\log'(y) &=& \frac{1}{y} \quad (\text{Substitute $y=e^x$} )
\end{alignat*}
$$

## $\log(xy) = \log(x) + \log(y)$
- Since $e^{a+b}=e^a e^b$, it follows that $\log(e^{a+b})=a+b$
- Since $\log(e^{a}) = a$ and $\log(e^{b})=b$, that means $\log(e^{a+b}) = \log(e^a) + \log(e^b)$
- But $\log(e^{a+b}) = \log(e^{a}e^{b}) = \log(e^a) + \log(e^b)$
- Substitute $e^a =x$ and $e^b =y$
- And $\log(xy) = \log(x) + \log(y)$

## Example: $\log$ and Optimization
- We often have an objective function that looks like this,
$$
f(x) = g_1(x) \times g_2(x) \times ... \times g_n(x) = \prod_{i=1}^n g_i(x)
$$
where $g_i(x)>0$, and we want to maximize over $x$
- Applying the multiplication rule here would be brutal
- Instead, we can convert the product into a summation:
$$
\ell(x) = \log(f(x)) = \sum_{i=1}^n \log(g_i(x))
$$
- And then this is much easier to analyze,
$$
\ell'(x) = \sum_{i=1}^n \frac{g_i '(x)}{g_i(x)}
$$
than this:
$$
f'(x) = \sum_{i=1}^n g'_i(x) \prod_{j \neq i} g_j(x)
$$

## Why?
- This is an example of a more general trick
- A transformation $t(y)$ is **increasing** if $y>y'$ implies $t(y)>t(y')$
- $x^*$ is a maximizer of $f$ if $f(x^*) \ge f(x')$ for all other $x'$
- Then 
$$
f(x^*) \ge f(x') \quad \longrightarrow \quad t(f(x^*)) \ge t(f(x'))
$$
- So we can always take increasing transformations of our objective functions to eliminate difficult features
